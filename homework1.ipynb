**(d) Create a pie chart that takes the total of AtBat in 1986 and displays what portion corresponded to Hits, Walks, or neither.**

```python
import numpy as np

#find total num of stats
total_atbats= df['AtBat'].sum()
num_hits = df['Hits'].sum()
num_walks = df['Walks'].sum()
neither = (total_atbats - num_hits - num_walks)/total_atbats

#set pie chart values
options = ['% Hits', '% Walks', '% Neither']
percents = [(num_hits/total_atbats), (num_walks/total_atbats), neither]

#create figure and axes
fig, ax = plt.subplots()

#create pie chart
ax.pie(percents, radius=3, center=(4,4), autopct='%.2f',
       wedgeprops={'linewidth': 1, 'edgecolor': 'white'}, frame =True)

#add ledgend
ax.legend(options, loc='right')

#adjust axes
ax.set(xlim=(0,12), xticks=np.arange(1,8),
       ylim=(0,9), yticks=np.arange(1,8))
ax.tick_params(left=False, labelleft=False,
               bottom=False, labelbottom=False)

ax.set(title='Pie chart of at bats in MLB during 1986')

#print pie chart
plt.show()
```

**(g) Make pairwise scatterplots for each of the 6 career metrics with Salary.**

```python
#coulmn names and labels
metrics = {
    "CAtBat": "Career at Bats",
    "CHits": "Career Hits",
    "CHmRun": "Career Home Runs",
    "CRuns": "Career Runs",
    "CRBI": "Career RBIs",
    "CWalks": "Career Walks"
}

#create subplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

#loop through metrics
for i, (col, label) in enumerate(metrics.items()):
    axes[i].scatter(df[col], df['Salary'], alpha=0.6, color='blue', edgecolors='black')
    axes[i].set_xlabel(label)
    axes[i].set_ylabel("Salary (in $1,000s)")
    axes[i].set_title(f"Salary vs {label}")

#display plots
plt.tight_layout()
plt.show()
```

**(h) Calculate the mean salary for each division. Which division spends more on salaries? How much? Answer in total dollars rather than a per-player average.**

```python
#seperate salaries by division
ESalarys = df.loc[df['Division'] == 'E', 'Salary']
WSalarys = df.loc[df['Division'] == 'W', 'Salary']

#calculate statistics
Emean = (ESalarys.mean()) * 1000
Wmean = (WSalarys.mean()) * 1000
Esum = (ESalarys.sum()) * 1000
Wsum = (WSalarys.sum()) * 1000

#print out statistics
print(f'Mean salary in division E: ${Emean:,.2f}')
print(f'Mean salary in division W: ${Wmean:,.2f}')
print(f'Total amount E division spent on salaries: ${Esum:,.2f}')
print(f'Total amount W division spent on salaries: ${Wsum:,.2f}')
```

As seen above, the E division spends the the most on salaries. To be exact, they spent $20,113,505 more than the W division.

**(i) Provide boxplots for Salary for players with tenure greater than 12 years versus less than 5 years. What can you conclude?**
