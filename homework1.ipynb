**(b) From what year(s) is this data? What is the difference between HmRun and CHmRun?**

From reading the details about the data, I learned that this data is from the 1986 and 1987 seasons. Also, HmRun is the number of home runs in 1986 while CHmRun is the number of homeruns during the player's career.

**(c) Find the median of each numeric variable.**

```python
#find median of stats and print them out
print('Median # of at bats in 1986:', df['AtBat'].median())
print('Median # of hits in 1986:', df['Hits'].median())
print('Median # of homeruns in 1986:', df['HmRun'].median())
print('Median # of runs in 1986:', df['Runs'].median())
print('Median # of runs batted in in 1986:', df['RBI'].median())
print('Median # of walks in 1986:', df['Walks'].median())
print('Median # of years in the major leagues:', df['Years'].median())
print('Median # of at bats in career:', df['CAtBat'].median())
print('Median # of hits in career:', df['CHits'].median())
print('Median # of home runs in career:', df['CHmRun'].median())
print('Median # of runs in career:', df['CRuns'].median())
print('Median # of runs batted in in career:', df['CRBI'].median())
print('Median # of walks in career:', df['CWalks'].median())
print('Median # of put outs in 1986:', df['PutOuts'].median())
print('Median # of assists in 1986:', df['Assists'].median())
print('Median # of errors in 1986:', df['Errors'].median())
print('Median salary on opening day in 1987 (in thousands of $):', df['Salary'].median())
```

**(d) Create a pie chart that takes the total of AtBat in 1986 and displays what portion corresponded to Hits, Walks, or neither.**

```python
import numpy as np

#find total num of stats
total_atbats= df['AtBat'].sum()
num_hits = df['Hits'].sum()
num_walks = df['Walks'].sum()
neither = (total_atbats - num_hits - num_walks)/total_atbats

#set pie chart values
options = ['% Hits', '% Walks', '% Neither']
percents = [(num_hits/total_atbats), (num_walks/total_atbats), neither]

#create figure and axes
fig, ax = plt.subplots()

#create pie chart
ax.pie(percents, radius=3, center=(4,4), autopct='%.2f',
       wedgeprops={'linewidth': 1, 'edgecolor': 'white'}, frame =True)

#add ledgend
ax.legend(options, loc='right')

#adjust axes
ax.set(xlim=(0,12), xticks=np.arange(1,8),
       ylim=(0,9), yticks=np.arange(1,8))
ax.tick_params(left=False, labelleft=False,
               bottom=False, labelbottom=False)

ax.set(title='Pie chart of at bats in MLB during 1986')

#print pie chart
plt.show()
```
